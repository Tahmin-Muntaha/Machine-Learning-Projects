```python
from sklearn.model_selection import train_test_split
X=df.iloc[:,1:]
y=df.iloc[:,0]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=43)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
```

```python
trf=ColumnTransformer(
        transformers=[
            ('ohe',OneHotEncoder(),['mainroad','guestroom','basement','hotwaterheating','airconditioning','prefarea','furnishingstatus'])
        ],
        remainder='passthrough'
    
)
pipe=Pipeline([('ohe',trf)])
```

```python
X_train=pipe.fit_transform(X_train)
X_test=pipe.transform(X_test)
```

```python
columns = pipe.get_feature_names_out()

X_train = pd.DataFrame(X_train, columns=columns, index=y_train.index)
X_test = pd.DataFrame(X_test, columns=columns, index=y_test.index)
```

```python
X_train.head()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score
```

```python
lr=LinearRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
r2_score(y_test,lr.predict(X_test))
```

```python
poly=PolynomialFeatures(degree=2)
```

```python
X_train_poly=poly.fit_transform(X_train)
X_test_poly=poly.transform(X_test)
```

```python
lr.fit(X_train_poly,y_train)
```

```python
y_pred=lr.predict(X_test_poly)
r2_score(y_test,y_pred)
```

```python
from sklearn.linear_model import SGDClassifier
```

```python
sgd=SGDClassifier()
```

```python
sgd.fit(X_train_poly,y_train)
```

```python
r2_score(y_test,sgd.predict(X_test_poly))
```
